Import the required libraries and load the Taxis dataset

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Taxis dataset

df = sns.load_dataset("taxis")

# Display first 10 rows

df.head(10)
```


## 3. Handling Missing Values

Check for missing values and handle them using
appropriate methods such as median and mode.


```python

df.isnull().sum()
```

```python
# Convert columns to datetime

df["pickup"] = pd.to_datetime(df["pickup"])
df["dropoff"] = pd.to_datetime(df["dropoff"])
df.info()
```


# Remove rows with missing pickup time or fare


```python

# Remove rows with missing pickup time or fare

df = df.dropna(subset=["pickup", "fare"])
```

```python

# Fill categorical columns using mode

categorical_cols = df.select_dtypes(
    include=["object", "category", "string"]
).columns

for col in categorical_cols:
    mode_value = df[col].mode()

    if not mode_value.empty:
        df[col] = df[col].fillna(mode_value[0])
```

```python

# Fill numerical columns using median

numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python

# Check missing values after cleaning

df.isnull().sum()
```

## 3. Visualizations using Matplotlib/Pandas

#### Line Chart
#### Plot a line chart to visualize the fare over time, using the pickup timestamp as the
#### x-axis and fare as the y-axis. Ensure the pickup column is converted to a datetime
#### format before plotting.

### 3.1 Line Chart – Fare Over Time

##### A line chart is used to visualize fare changes over time.
##### Here, pickup timestamp is used on the x-axis and fare on the y-axis.

```python
# Sort data by pickup time
df = df.sort_values("pickup")

# Plot line chart
plt.figure(figsize=(6, 3))
plt.plot(df["pickup"], df["fare"])

plt.title("Fare Over Time")
plt.xlabel("Pickup Date and Time")
plt.ylabel("Fare ($)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##### sort_values() sorts the pickup timestamps.
##### plt.plot() creates the line chart.
##### X-axis shows pickup date and time.
##### Y-axis shows taxi fare.

### 3.2 Bar Chart – Total Fare by Pickup Borough

##### A bar chart is used to compare values across categories.
##### Here, I will show the total fare collected from each pickup borough.

```python
# Calculate total fare by pickup borough
borough_fare = df.groupby("pickup_borough")["fare"].sum()

# Plot bar chart
plt.figure(figsize=(5, 2.5))
borough_fare.plot(kind="bar")

plt.title("Total Fare by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare ($)")

plt.xticks(rotation=45)
plt.show()
```

##### groupby() groups the data by pickup borough.
##### sum() calculates the total fare for each borough.
##### kind="bar" creates a bar chart.

### 3.3 Pie Chart – Trips by Payment Method

##### A pie chart shows the distribution of data across different categories.
##### Here, I will show the number of trips made using each payment method.

```python
# Count trips by payment method
payment_counts = df["payment"].value_counts()

# Plot pie chart
plt.figure(figsize=(5, 5))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%"
)

plt.title("Trips by Payment Method")
plt.show()
```

##### value_counts() counts trips for each payment method.
##### labels displays payment method names.
##### autopct displays the percentage of each category.

### 3.4 Histogram – Distribution of Distance

##### A histogram is used to show the distribution of numerical data.
##### Here, I will create a histogram to visualize the distribution of trip distances. I will use 20 bins to get a better
##### understanding of how trip distances are spread.

```python
plt.figure(figsize=(6, 3))

plt.hist(df['distance'], bins=20, edgecolor='black')

plt.title('Distribution of Trip Distance')
plt.xlabel('Distance')
plt.ylabel('Number of Trips')

plt.show()
```

##### plt.figure(figsize=(8, 5)) – Sets the chart size.
##### plt.hist() – Creates the histogram.
##### df['distance'] – Uses the distance column.
##### bins=20 – Divides the distance values into 20 groups.
##### edgecolor='black' – Adds borders to the bars.
##### plt.show() – Displays the chart.

### 3.5 Box Plot – Tip Amount by Pickup Borough

##### A box plot is used to understand the distribution of numerical data across different categories.
##### Here, I will create a box plot to compare the distribution of tip amounts across different pickup boroughs.

```python
plt.figure(figsize=(6, 3))

df.boxplot(column='tip', by='pickup_borough')

plt.title('Tip Amount by Pickup Borough')
plt.suptitle('')
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')

plt.show()
```

##### plt.figure() – Sets the chart size.
##### df.boxplot() – Creates the box plot.
##### column='tip' – Uses tip amounts as numerical data.
##### by='pickup_borough' – Groups the tips by pickup borough.
##### plt.suptitle('') – Removes the default extra title.
##### plt.show() – Displays the chart.

## 4. Visualizations using Seaborn

##### Seaborn is a Python library used to create attractive and informative statistical visualizations.

### 4.1 Count Plot – Trips by Pickup Borough

##### A count plot is used to display the number of observations in each category.
##### Here, I will create a count plot to visualize the number of trips in each pickup borough.

```python
import seaborn as sns

plt.figure(figsize=(5, 2.5))

sns.countplot(data=df, x='pickup_borough')

plt.title('Number of Trips by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Number of Trips')

plt.show()
```

##### sns.countplot() – Creates a count plot.
##### data=df – Uses our dataset.
##### x='pickup_borough' – Displays boroughs on the X-axis.
##### The Y-axis automatically shows the number of trips.
##### plt.title() – Adds the chart title.
##### plt.show() – Displays the chart.

### 4.2 Scatter Plot – Distance vs Fare

##### A scatter plot is used to show the relationship between two numerical variables.

##### Here, I will create a scatter plot to visualize the relationship between trip distance and fare.
##### I will use different colors to represent pickup boroughs.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='distance', y='fare', hue='pickup_borough')

plt.title('Distance vs Fare by Pickup Borough')
plt.xlabel('Distance')
plt.ylabel('Fare')

plt.show()
```

##### sns.scatterplot() – Creates a scatter plot.
##### x='distance' – Shows trip distance on the X-axis.
##### y='fare' – Shows fare on the Y-axis.
##### hue='pickup_borough' – Uses different colors for each borough.
##### plt.show() – Displays the chart.

### 4.3 Heatmap – Correlation Between Numerical Variables

##### A heatmap is used to visualize the correlation between numerical variables using different colors.
##### Here, I will create a correlation matrix to understand the relationships between distance, fare, tip, tolls, and total.

```python
# Select numerical columns
numeric_data = df[['distance', 'fare', 'tip', 'tolls', 'total']]

# Create correlation matrix
correlation_matrix = numeric_data.corr()

# Plot heatmap
plt.figure(figsize=(6, 4))

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap of Trip Variables')

plt.show()
```

##### numeric_data – Selects the five numerical columns.
##### .corr() – Calculates the correlation between the columns.
##### sns.heatmap() – Creates the heatmap.
##### annot=True – Displays correlation values inside the boxes.
##### cmap='coolwarm' – Uses different colors to represent correlations.
##### fmt='.2f' – Displays values up to two decimal places.
##### Close to +1 – Strong positive relationship.
##### Close to -1 – Strong negative relationship.
##### Close to 0 – Weak or no linear relationship.

### 4.4 Pair Plot – Pairwise Relationships

##### A pair plot is used to visualize the relationships between multiple numerical variables.
##### Here, I will create a pair plot to compare distance, fare, tip, and total.
##### I will use different colors based on the pickup zone to compare trips across different zones.

```python
sns.pairplot(df.sample(200), vars=['distance', 'fare', 'tip', 'total'], hue='pickup_zone')

plt.show()
```

##### sns.pairplot() – Creates pairwise plots between numerical variables.
##### vars – Specifies the numerical columns to visualize.
##### hue='pickup_zone' – Uses different colors for different pickup zone.
##### plt.show() – Displays the pair plot.
